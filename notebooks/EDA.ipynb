# Exploratory Data Analysis - Student Marks Prediction

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '..')
from src.utils import generate_dataset, display_dataset_info
```

## 1. Generate & Load Dataset

```python
df = generate_dataset()
display_dataset_info(df)
```

## 2. Scatter Plot

```python
plt.figure(figsize=(8, 6))
plt.scatter(df['Study_Hours'], df['Marks'], color='steelblue', alpha=0.7, edgecolors='white', s=60)
plt.title('Study Hours vs Marks', fontsize=14, fontweight='bold')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Histograms

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].hist(df['Study_Hours'], bins=15, color='steelblue', edgecolor='white')
axes[0].set_title('Study Hours Distribution')
axes[1].hist(df['Marks'], bins=15, color='coral', edgecolor='white')
axes[1].set_title('Marks Distribution')
plt.tight_layout()
plt.show()
```

## 4. Box Plots

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].boxplot(df['Study_Hours'], patch_artist=True, boxprops=dict(facecolor='steelblue'))
axes[0].set_title('Study Hours')
axes[1].boxplot(df['Marks'], patch_artist=True, boxprops=dict(facecolor='coral'))
axes[1].set_title('Marks')
plt.tight_layout()
plt.show()
```

## 5. Correlation Heatmap

```python
print('Correlation Matrix:')
print(df.corr())
print(f'\nCorrelation: {df["Study_Hours"].corr(df["Marks"]):.4f}')
```
